```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# 1. Dataset
dach_length = [55, 57, 64, 63, 58, 49, 54, 61]
dach_height = [30, 31, 36, 30, 33, 25, 37, 34]

jin_length = [56, 47, 56, 46, 49, 53, 52, 48]
jin_height = [52, 52, 50, 53, 50, 53, 49, 54]

# Features: [Length, Height], Labels: Dachshund = 0, Jindo = 1
X = np.array(
    [
        [l, h]
        for l, h in zip(
            dach_length + jin_length, dach_height + jin_height
        )
    ]
)
y = np.array([0] * len(dach_length) + [1] * len(jin_length))

# 2. Train Logistic Regression
model = LogisticRegression()
model.fit(X, y)

w1, w2 = model.coef_[0]
b = model.intercept_[0]

print(f"Model parameters: w1={w1:.4f}, w2={w2:.4f}, b={b:.4f}")

# 3. Calculate Decision Boundary Line
x_boundary = np.linspace(42, 68, 100)
y_boundary = -(w1 * x_boundary + b) / w2

# 4. Visualization
plt.figure(figsize=(9, 6))

# Data points
plt.scatter(
    dach_length, dach_height, c="crimson", label="Dachshund (0)", s=70
)
plt.scatter(
    jin_length,
    jin_height,
    c="royalblue",
    marker="^",
    label="Jindo dog (1)",
    s=70,
)

# Decision boundary
plt.plot(
    x_boundary,
    y_boundary,
    color="black",
    linestyle="--",
    linewidth=2,
    label="Decision Boundary ($P = 0.5$)",
)

# Styling
plt.title(
    "Dachshund vs. Jindo Classification (Logistic Regression)", fontsize=13
)
plt.xlabel("Length (cm)", fontsize=11)
plt.ylabel("Height (cm)", fontsize=11)
plt.xlim(42, 68)
plt.ylim(20, 60)
plt.legend(loc="upper left")
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()

# 5. Predict on New Data
new_dogs = np.array(
    [
        [58, 32],  # Expected: Dachshund (long body, short legs)
        [50, 52],  # Expected: Jindo dog (taller, shorter body)
    ]
)

predictions = model.predict(new_dogs)
probabilities = model.predict_proba(new_dogs)

labels = {0: "Dachshund", 1: "Jindo dog"}
for dog, pred, prob in zip(new_dogs, predictions, probabilities):
    print(
        f"Dog [Length: {dog[0]}, Height: {dog[1]}] -> {labels[pred]} "
        f"(Prob: Dachshund {prob[0]:.2%}, Jindo {prob[1]:.2%})"
    )
```
